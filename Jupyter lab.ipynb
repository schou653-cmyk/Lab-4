```python
from classes1 import Team, Driver

teams = {}

# Task 2: Load the driver and team information from the CSV
with open('f1_points.csv', 'r') as file:
    next(file)  # Skip the CSV header row

    for line in file:
        data = line.strip().split(',')

        driver = Driver(data[0], int(data[2]))
        team_name = data[1]

        if team_name not in teams:
            teams[team_name] = Team(team_name)

        teams[team_name].add_driver(driver)

print(f'number of teams: ', len(teams))
```

```python
#Verify Drivers and Team Totals

for team in teams.values():
    print(team)
    print()
```

```python
#Verify the Driver Objects

redbull = teams["RED BULL RACING HONDA RBPT"]

print("Team", redbull.name)
print("Drivers:", redbull.drivers)
```

```python
# Invaled DATA
import csv

invalid_data = [

    ["Driver A", "Test Team", "100"],
    ["Driver B", "Test team", "abc"]
]

with open ("invalid_f1.csv", "w", newline='') as file:
    writer = csv.writer(file)
    writer.writerow(["driver", "Team", "Points"])
    writer.writerows(invalid_data)

print("Invalid test file created")
```

```python
try:
    test_teams = {}

    with open("invalid_f1.csv", 'r') as file:
        reader = csv.reader(file)
        next(reader)

        for row in reader:
            driver_name = row[0]
            team_name = row[1]
            points = int(row[2])

            if team_name not in test_teams:
                test_teams[team_name] = Team(team_name)

            driver = Driver(driver_name, points)
            test_teams[team_name].add_driver(driver)


except ValueError as e:

    print("Error detected: ", e)
```
